# Python Learning Log: testing the sample programs

Every log entry says how its program was tested. This notebook shows that habit on the four fictional sample entries in `src/sample_entries/`: run each program with values you can check by hand, including exact boundary values, and compare.

The sample programs ask for input with `input()`. Here the typed values are passed in as text, which is the same as typing them at the prompt, so the prompts and results print on one line.

```python
from pathlib import Path
import subprocess
import sys

# Works whether the notebook runs from notebooks/ or from the repository root.
candidates = [Path.cwd().parent / "src" / "sample_entries", Path.cwd() / "src" / "sample_entries"]
SAMPLES = next(folder for folder in candidates if folder.is_dir())


def run_entry(filename, typed=""):
    """Run one sample program, typing each value in `typed` at its prompts."""
    result = subprocess.run([sys.executable, str(SAMPLES / filename)], input=typed,
                            text=True, capture_output=True, check=True)
    return result.stdout.strip()


print(sorted(path.name for path in SAMPLES.glob("*.py")))
```

## Entry 1: temperature conversion

Two values with well-known answers: water freezes at 0 °C (32 °F) and boils at 100 °C (212 °F).

```python
for celsius in [0, 100, -40]:
    print(run_entry("week01_entry01_temperature_converter.py", f"{celsius}\n"))
    print("  hand check:", celsius * 9 / 5 + 32)
```

−40 is a useful extra test: it is the one temperature that is the same on both scales.

## Entry 2: grade classification

The sample narrative's bug was at the cutoffs, so test exactly at each boundary and just below it.

```python
for score in [90, 89.9, 80, 79.9, 70, 60, 59.9]:
    print(score, "->", run_entry("week01_entry02_grade_classifier.py", f"{score}\n").split("Grade: ")[1])
```

## Entry 3: average calculator

Five values whose mean you can see at a glance, then decimal values to confirm that `float()` accepts them.

```python
for values in [[10, 20, 30, 40, 50], [1.5, 2.5, 3.5, 4.5, 5.5]]:
    typed = "".join(f"{value}\n" for value in values)
    print(run_entry("week01_entry03_average_calculator.py", typed).split("Enter number 5: ")[1])
    print("  hand check:", sum(values) / len(values))
```

## Entry 4: data summary

This program needs no input. Check its results against the same calculation done separately.

```python
print(run_entry("week01_entry04_data_summary.py"))

data = [12, 18, 25, 30, 22, 16, 19]
print("  hand check: min", min(data), "max", max(data), "sum", sum(data), "n", len(data),
      "average", round(sum(data) / len(data), 2))
```
